Accept 5 paper marks, roll no.,name;
and calculate total and percentage
if percentage >= 85 , student can take admission
else can't take admission
using class and object and constructor

```python
class Student:
    def __init__(self):
        self.roll_no = input("Enter Roll Number: ")
        self.name = input("Enter Student Name: ")


        self.marks = []
        print("Enter marks for 5 papers:")
        for i in range(5):
            mark = float(input(f"Paper {i + 1}: "))
            self.marks.append(mark)


        self.total = sum(self.marks)
        self.percentage = (self.total / 500) * 100

    def display_result(self):


        print("<-----STUDENT RESULT----->")

        print(f"Roll No    : {self.roll_no}")
        print(f"Name       : {self.name}")
        print(f"Total Marks: {self.total} / 500")
        print(f"Percentage : {self.percentage:.2f}%")
        print("--------------------------")


        if self.percentage >= 25:
            print("Status     : Student can take admission.")
        else:
            print("Status     : Student can't take admission.")


student_object = Student()
student_object.display_result()
```

Single level Inheritance

```python
class A:
  def showA(self):
    print("showA function")


class B(A):
  def showB(self):
    print("showB function")


class C(B):
  def showC(self):
    print("showC function")

obj=C()
obj.showA()
obj.showB()
obj.showC()
```

```python
class A:
  def showA(self):
    print("showA function")


class B:
  def showB(self):
    print("showB function")


class C(A,B):
  def showC(self):
    print("showC function")

obj=C()
obj.showA()
obj.showB()
obj.showC()
```

```python
class D:
    def showA(self):
        print("showA function")


class A(D):
    def showB(self):
        print("showB function")


class B(D):
    def showC(self):
        print("showC function")


class C(D):
    def showD(self):
        print("showD function")


class E(A, B, C):
    pass


obj = E()

obj.showA()
obj.showB()
obj.showC()
obj.showD()
```
