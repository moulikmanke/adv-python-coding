```python
import pandas as pd

# Creating the dictionary with data
data = {
    'Name': ['A', 'B', 'C', 'D', 'E'],
    'Math':[80, 78, 92, 65, 88],
    'Science':[95, 89, 91, 95, 998],
    'English': [85, 76, 92, 85, 88]
}

# Converting the dictionary into a pandas DataFrame
df = pd.DataFrame(data)

# Displaying the DataFrame
print(df)
```

```python
import numpy as np
import pandas as pd

# Extracting values from specific columns
marks = df[['Math', 'Science', 'English']].values

# Calculating the row-wise mean (axis=1) for each student
average_marks = np.mean(marks, axis=1)

# Adding the calculated averages as a new column to the DataFrame
df['Average'] = average_marks

# Displaying the updated DataFrame
print(df)
```
